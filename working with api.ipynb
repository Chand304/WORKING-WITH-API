```python
df = pd.DataFrame()

for i in range(1, 569):

    params = {
        "api_key": "https://api.themoviedb.org/3/movie/top_rated?api_key=94ce7342958db89ec0cb3c63589e6255",
        "page": i
    }

    response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=94ce7342958db89ec0cb3c63589e6255", params=params)

    if response.status_code != 200:
        print("Error page:", i)
        print(response.json())
        break

    temp_df = pd.DataFrame(response.json()['results'])

    temp_df = temp_df[
        ['adult', 'id', 'release_date', 'title',
         'popularity', 'vote_average']
    ]

    df = pd.concat([df, temp_df], ignore_index=True)

print("Total rows:", len(df))
```
